# 🦥 Unsloth LoRA Studio: Merge, Quantize & Publish

**Description:** interactive notebook to merge LoRA adapters with base LLMs, quantize them into multiple GGUF formats simultaneously, and push them to Hugging Face. Intened for use in Google Colab.

### ⚡ Features
* **Interactive Dashboard:** No more editing code variables. Use dropdowns and buttons.
* **Multi-Quantization:** Select multiple formats (e.g., Q4_K_M, Q8_0) and process them in one go.
* **Smart Upload:** Drag-and-drop your LoRA zip file.

```python
# @title 1. Initialize Environment
# @markdown Run this cell to install Unsloth and prepare the dashboard dependencies.
import torch
try:
    import unsloth
except ImportError:
    major_version, minor_version = torch.cuda.get_device_capability()
    if major_version >= 8:
        !pip install "unsloth[colab-ampere] @ git+https://github.com/unslothai/unsloth.git" --quiet
    else:
        !pip install "unsloth[colab] @ git+https://github.com/unslothai/unsloth.git" --quiet
    !pip install --no-deps "xformers<0.0.27" "trl<0.8.6" peft accelerate bitsandbytes --quiet
    !pip install llama-cpp-python --quiet

import ipywidgets as widgets
from IPython.display import display, clear_output
import os
import shutil
from huggingface_hub import login
from unsloth import FastLanguageModel

print("✅ Environment Ready. Run the next cell to open the Studio.")
```

```python
# @title 2. The Unsloth Studio Dashboard
# @markdown Configure your model, upload your LoRA, and select your export options below.

# --- Widget Definitions ---
style = {'description_width': '120px'}
layout = widgets.Layout(width='95%')

# Header
header = widgets.HTML("<h2>🎛️ Model Configuration</h2>")

# Hugging Face Token
hf_token_widget = widgets.Password(
    value='',
    placeholder='hf_...',
    description='HF Write Token:',
    style=style, layout=layout
)

# Base Model Selection (Combobox for searchability)
base_model_widget = widgets.Combobox(
    placeholder='Start typing (e.g., unsloth/llama-3-8b...)',
    options=[
        'unsloth/llama-3-8b-bnb-4bit',
        'unsloth/llama-3-8b-Instruct-bnb-4bit',
        'unsloth/mistral-7b-bnb-4bit',
        'unsloth/gemma-7b-bnb-4bit',
        'unsloth/gemma-2b-bnb-4bit',
        'unsloth/phi-3-mini-4k-instruct',
    ],
    description='Base Model:',
    ensure_option=False,
    value='unsloth/llama-3-8b-bnb-4bit',
    style=style, layout=layout
)

# File Upload
upload_widget = widgets.FileUpload(
    accept='.zip',
    multiple=False,
    description='Upload LoRA .zip',
    layout=widgets.Layout(width='30%')
)
upload_status = widgets.Label(value="Waiting for upload...")

# Quantization Selection
quant_widget = widgets.SelectMultiple(
    options=['q4_k_m', 'q5_k_m', 'q8_0', 'f16'],
    value=['q4_k_m'],
    rows=4,
    description='Quant Methods:',
    style=style, layout=layout
)

# Export Names
username_widget = widgets.Text(
    placeholder='Your HF Username',
    description='HF Username:',
    style=style, layout=layout
)
new_model_name_widget = widgets.Text(
    value='my-finetuned-model-gguf',
    description='New Model Name:',
    style=style, layout=layout
)

# Run Button
run_btn = widgets.Button(
    description='🚀 Start Merge & Upload',
    button_style='success', # 'success', 'info', 'warning', 'danger' or ''
    layout=widgets.Layout(width='100%', height='50px'),
    icon='rocket'
)

# Output Log
out = widgets.Output()

# --- Logic ---
def on_click_run(b):
    out.clear_output()
    with out:
        # 1. Validation
        if not hf_token_widget.value:
            print("❌ Error: Please enter a Hugging Face Token.")
            return
        if not upload_widget.value:
            print("❌ Error: Please upload a LoRA .zip file.")
            return
        
        # 2. Login
        print("🔑 Logging into Hugging Face...")
        login(token=hf_token_widget.value)

        # 3. Handle File Upload
        print("📂 Processing LoRA file...")
        os.makedirs("lora_upload", exist_ok=True)
        
        # Ipwyidgets file upload value format varies by version
        vals = upload_widget.value
        # Standardizing access to content
        if isinstance(vals, dict): # older ipywidgets
             fname = next(iter(vals))
             content = vals[fname]['content']
        else: # newer ipywidgets (tuple)
             file_info = vals[0]
             fname = file_info['name']
             content = file_info['content']
             
        with open(fname, "wb") as f:
            f.write(content)
        
        import zipfile
        with zipfile.ZipFile(fname, 'r') as zip_ref:
            zip_ref.extractall("lora_upload")
        
        # Find extracted folder
        lora_path = "lora_upload"
        for root, dirs, files in os.walk("lora_upload"):
            if "adapter_config.json" in files:
                lora_path = root
                break
        print(f"   Found adapter at: {lora_path}")

        # 4. Load Model
        print(f"⏳ Loading Base Model: {base_model_widget.value}...")
        model, tokenizer = FastLanguageModel.from_pretrained(
            model_name = base_model_widget.value,
            max_seq_length = 2048,
            dtype = None,
            load_in_4bit = True,
        )

        print("🔗 Attaching LoRA...")
        model = FastLanguageModel.get_peft_model(
            model,
            r = 16,
            target_modules = ["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"],
            lora_alpha = 16,
            lora_dropout = 0,
            bias = "none",
            use_gradient_checkpointing = False,
            random_state = 3407,
        )
        model.load_adapter(lora_path)

        # 5. Quantize and Upload Loop
        selected_quants = quant_widget.value
        repo_id = f"{username_widget.value}/{new_model_name_widget.value}"
        
        print(f"🚀 Starting Batch Processing for: {selected_quants}")
        print(f"   Target Repo: https://huggingface.co/{repo_id}")

        for method in selected_quants:
            print(f"\n--- Processing {method} ---")
            try:
                if method == "f16":
                     model.push_to_hub_merged(
                        repo_id,
                        tokenizer,
                        save_method = "merged_16bit",
                        token = hf_token_widget.value,
                    )
                else:
                    model.push_to_hub_gguf(
                        repo_id,
                        tokenizer,
                        quantization_method = method,
                        token = hf_token_widget.value,
                    )
                print(f"✅ {method} Uploaded Successfully!")
            except Exception as e:
                print(f"❌ Failed to upload {method}: {e}")
        
        print("\n✨ All tasks finished!")

run_btn.on_click(on_click_run)

# Layout
ui = widgets.VBox([
    header,
    hf_token_widget,
    widgets.HTML("<hr>"),
    base_model_widget,
    widgets.HBox([upload_widget, upload_status]),
    widgets.HTML("<hr>"),
    widgets.HBox([username_widget, new_model_name_widget]),
    quant_widget,
    widgets.HTML("<br>"),
    run_btn,
    widgets.HTML("<hr>"),
    out
])

display(ui)
```
